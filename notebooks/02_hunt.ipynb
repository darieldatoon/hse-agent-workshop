# 02 · The hunt

The agent you just built has been triaging incident reports at Juniper Basin Energy's sites for
a week. Its traces are in your workspace. **Something is wrong with it, and more than one
thing.** Find out what.

There's no code in this lesson. Everything happens in LangSmith, so if Colab gave you trouble,
you can still do all of it.

---

## 1. Open the traces

1. Open [smith.langchain.com](https://smith.langchain.com). If you're not in your workspace yet:
   click **your name** at the bottom left, select the **workshop organization**, then the
   workspace with **your name**.
2. Open **Tracing** and select the project **`hse-triage-prod`**. Each row is one report the
   agent triaged.
3. Click a row to open its trace: the report at the top (with its photo, if it has one), then
   every model and tool call, then the `Triage` the agent settled on.

---

## 2. Five ways to look

Each of these can turn something up. Try them all.

- **Look at the photos.** About one report in four has one. Does the output match what you see?
- **Compare the report with the output.** Is everything in the output actually in the report?
- **Open the steps.** Which tools did the agent call? Did it check everything it should have?
- **Sort by tokens.** Which runs cost the most, and why?
- **Read the feedback.** Some reporters rated the agent's answer 👍 or 👎. Do you agree with
  them?

---

## 3. Flag what you find

When a trace is wrong:

1. In the trace, add feedback with the key **`flag`** and a score of **1**.
2. In the comment, say in a few words what's wrong.

You don't need to flag everything. Aim for a few examples of each kind of problem you find:
the traces you flag become your dataset in the next lesson.

---

## 4. Debrief

When time is up, we'll name the problems together. Be ready to say:

- What kinds of problems did you find?
- How did you spot each one?
- Which would you have caught without reading traces?

---

## 📌 Key takeaways

- A trace shows what the agent actually did, not what you expected it to do.
- Some problems show up in a single trace. Others only show up across many, such as cost.
- Feedback from users isn't a measure of correctness. A 👍 can sit on a wrong answer.
- Everything you found by reading, you can turn into a check that runs on every version. That's
  next.
